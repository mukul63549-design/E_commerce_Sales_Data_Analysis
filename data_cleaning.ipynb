```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv("train.csv")
```

```python
data.duplicated().sum()
```

```python
data = data.drop_duplicates()
```

```python
data.isnull().sum()
```

```python
data["Postal Code"] = data["Postal Code"].fillna(
    data["Postal Code"].median())
```

```python
data["Postal Code"].isna().sum()
```

```python
data.isna().sum().sum()
```

```python
(data.isnull().sum() / len(data) * 100).sort_values(ascending=False)
```

```python
data["Order Date"] = pd.to_datetime(
    data["Order Date"],
    errors="coerce"
)
```

```python
data["Ship Date"] = pd.to_datetime(
    data["Ship Date"],
    errors="coerce"
)
```

```python
data[["Order Date", "Ship Date"]].dtypes
```

```python
data["Postal Code"].dtype
```

```python
numeric_columns = [
    "Sales",
    "Postal Code"
]
```

```python
for col in numeric_columns:
    data[col] = pd.to_numeric(
        data[col],
        errors="coerce"
    )
```

```python
data[numeric_columns].dtypes
```

```python
data[numeric_columns].isnull().sum()
```

```python
for col in ["Sales"]:
    data[col] = data[col].fillna(
        data[col].median()
    )
```

```python
text_columns = data.select_dtypes(
    include="object"
).columns

text_columns
```

```python
for col in text_columns:
    data[col] = data[col].str.strip()
```

```python
data[text_columns].isnull().sum()
```

```python
for col in text_columns:
    data[col] = data[col].fillna("Unknown")
```

```python
(data["Sales"] < 0).sum()
```

```python
print("Invalid Order Dates:",
      data["Order Date"].isnull().sum())

print("Invalid Ship Dates:",
      data["Ship Date"].isnull().sum())
```

```python
(data["Ship Date"] < data["Order Date"]).sum()
```
